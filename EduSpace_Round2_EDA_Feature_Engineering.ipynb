# EduSpace Campus Services — Round 2


```python


# STEP 1: Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

np.random.seed(42)
sns.set_theme(style="whitegrid")
OUT = Path("eduspace_round2_output")
OUT.mkdir(exist_ok=True)

# STEP 2: Generate realistic synthetic source CSV files
facilities = pd.DataFrame({
    "facility": ["C101", "C102", "LAB1", "LAB2", "LIBRARY", "AUDITORIUM",
                 "SEMINAR1", "C201", "C202", "SPORTS HALL"],
    "type": ["Classroom", "Classroom", "Laboratory", "Laboratory", "Library",
             "Auditorium", "Seminar Room", "Classroom", "Classroom", "Sports"],
    "capacity": [60, 45, 35, 40, 120, 300, 50, 60, 45, 100],
    "block": ["A", "A", "B", "B", "C", "C", "B", "A", "A", "D"]
})
facilities.to_csv(OUT / "facilities.csv", index=False)

dates = pd.date_range("2026-01-05", periods=56, freq="D")
working_days = []
for d in dates:
    is_working = d.weekday() < 5
    working_days.append({
        "date": d.date().isoformat(),
        "working_day": is_working,
        "event": np.random.choice(["None", "Guest Lecture", "Workshop", "Cultural Event"], p=[.78,.08,.08,.06]),
        "exam_period": bool(pd.Timestamp("2026-02-02") <= d <= pd.Timestamp("2026-02-13"))
    })
calendar = pd.DataFrame(working_days)
calendar.to_csv(OUT / "calendar.csv", index=False)

# Bookings include deliberate duplicates and inconsistent facility labels.
booking_rows = []
purposes = ["Lecture", "Practical", "Study", "Meeting", "Workshop", "Exam"]
for d in dates:
    if d.weekday() < 5:
        for fac in facilities["facility"]:
            n = np.random.poisson(1.4)
            for _ in range(n):
                start_hour = int(np.random.choice([8,9,10,11,12,13,14,15,16], p=[.08,.13,.16,.13,.10,.12,.12,.10,.06]))
                duration = int(np.random.choice([1,2,3], p=[.55,.38,.07]))
                booking_rows.append({
                    "facility": fac,
                    "date": d.date().isoformat(),
                    "start": f"{start_hour:02d}:00",
                    "end": f"{min(start_hour+duration,18):02d}:00",
                    "purpose": np.random.choice(purposes, p=[.40,.22,.15,.08,.08,.07])
                })
bookings = pd.DataFrame(booking_rows)
if len(bookings) > 10:
    # add duplicate records
    bookings = pd.concat([bookings, bookings.sample(12, random_state=42)], ignore_index=True)
# Add inconsistent labels on a small subset
alias_map = {"C101":"c101 ", "LIBRARY":"Library", "LAB1":"Lab 1", "SPORTS HALL":"Sports hall"}
for old, new in alias_map.items():
    idx = bookings.index[bookings["facility"] == old]
    if len(idx):
        bookings.loc[idx[:max(1, len(idx)//12)], "facility"] = new
bookings.to_csv(OUT / "bookings.csv", index=False)

# Occupancy sampled every hour on working and weekend days.
occupancy_rows = []
for d in dates:
    for _, f in facilities.iterrows():
        for hour in range(8, 18):
            # baseline varies by facility type, hour, day, and exam period
            base = {"Classroom": .48, "Laboratory": .57, "Library": .40,
                    "Auditorium": .20, "Seminar Room": .35, "Sports": .30}[f["type"]]
            peak = .22 if hour in [10, 11, 14, 15] else 0
            exam_boost = .20 if (pd.Timestamp("2026-02-02") <= d <= pd.Timestamp("2026-02-13")) and f["type"] == "Library" else 0
            weekend_factor = .35 if d.weekday() >= 5 else 1
            ratio = max(0, base + peak + exam_boost + np.random.normal(0, .16)) * weekend_factor
            count = int(round(ratio * f["capacity"]))
            # Deliberately create some over-capacity readings
            if np.random.random() < .012:
                count = int(f["capacity"] * np.random.uniform(1.05, 1.25))
            # Deliberately create missing readings
            if np.random.random() < .035:
                count = np.nan
            occupancy_rows.append({
                "timestamp": f"{d.date().isoformat()} {hour:02d}:00:00",
                "facility": f["facility"],
                "count": count
            })
occupancy = pd.DataFrame(occupancy_rows)
occupancy.to_csv(OUT / "occupancy.csv", index=False)

# Maintenance data includes incomplete records.
maintenance_rows = []
for fac in facilities["facility"]:
    for _ in range(np.random.poisson(.65)):
        maintenance_rows.append({
            "facility": fac,
            "issue": np.random.choice(["Projector failure", "AC repair", "Network issue", "Electrical fault", "Furniture repair"]),
            "downtime": np.random.choice([1, 2, 3, 4, 6, np.nan], p=[.20,.25,.20,.15,.12,.08])
        })
maintenance = pd.DataFrame(maintenance_rows)
if len(maintenance):
    maintenance.loc[maintenance.sample(frac=.12, random_state=7).index, "issue"] = np.nan
maintenance.to_csv(OUT / "maintenance.csv", index=False)

print("Generated CSV files in:", OUT.resolve())
for filename in ["facilities.csv", "bookings.csv", "occupancy.csv", "calendar.csv", "maintenance.csv"]:
    print(f"{filename}: {len(pd.read_csv(OUT / filename))} rows")

# STEP 3: Load all CSV files and inspect structure
facilities = pd.read_csv(OUT / "facilities.csv")
bookings = pd.read_csv(OUT / "bookings.csv")
occupancy = pd.read_csv(OUT / "occupancy.csv")
calendar = pd.read_csv(OUT / "calendar.csv")
maintenance = pd.read_csv(OUT / "maintenance.csv")

print("\n--- FACILITIES ---")
display(facilities.head())
print("\n--- BOOKINGS ---")
display(bookings.head())
print("\n--- OCCUPANCY ---")
display(occupancy.head())
print("\n--- CALENDAR ---")
display(calendar.head())
print("\n--- MAINTENANCE ---")
display(maintenance.head())

# STEP 4: Data quality audit before cleaning
datasets = {
    "facilities": facilities, "bookings": bookings, "occupancy": occupancy,
    "calendar": calendar, "maintenance": maintenance
}
audit_rows = []
for name, df in datasets.items():
    audit_rows.append({
        "dataset": name,
        "rows": len(df),
        "columns": len(df.columns),
        "missing_cells": int(df.isna().sum().sum()),
        "duplicate_rows": int(df.duplicated().sum())
    })
audit = pd.DataFrame(audit_rows)
print("Data quality audit:")
display(audit)

print("Missing values by dataset:")
for name, df in datasets.items():
    print(f"\n{name}")
    display(df.isna().sum().to_frame("missing_count"))

# STEP 5: Clean data
# Standardize facility labels consistently
facility_aliases = {
    "c101": "C101", "c102": "C102", "lab 1": "LAB1", "lab1": "LAB1",
    "lab 2": "LAB2", "lab2": "LAB2", "library": "LIBRARY",
    "auditorium": "AUDITORIUM", "seminar1": "SEMINAR1",
    "c201": "C201", "c202": "C202", "sports hall": "SPORTS HALL"
}
def standardize_facility(series):
    return series.astype("string").str.strip().str.lower().map(facility_aliases).fillna(series.astype("string").str.strip().str.upper())

facilities["facility"] = standardize_facility(facilities["facility"])
bookings["facility"] = standardize_facility(bookings["facility"])
occupancy["facility"] = standardize_facility(occupancy["facility"])
maintenance["facility"] = standardize_facility(maintenance["facility"])

# Remove duplicate bookings
booking_rows_before = len(bookings)
bookings = bookings.drop_duplicates(subset=["facility", "date", "start", "end", "purpose"]).copy()
print("Duplicate booking rows removed:", booking_rows_before - len(bookings))

# Convert data types
bookings["date"] = pd.to_datetime(bookings["date"], errors="coerce")
bookings["start"] = pd.to_datetime(bookings["start"], format="%H:%M", errors="coerce").dt.hour
bookings["end"] = pd.to_datetime(bookings["end"], format="%H:%M", errors="coerce").dt.hour
occupancy["timestamp"] = pd.to_datetime(occupancy["timestamp"], errors="coerce")
occupancy["count"] = pd.to_numeric(occupancy["count"], errors="coerce")
calendar["date"] = pd.to_datetime(calendar["date"], errors="coerce")
calendar["working_day"] = calendar["working_day"].astype(bool)
calendar["exam_period"] = calendar["exam_period"].astype(bool)
maintenance["downtime"] = pd.to_numeric(maintenance["downtime"], errors="coerce")

# Remove invalid booking times and invalid timestamps
bookings = bookings.dropna(subset=["facility", "date", "start", "end"])
bookings = bookings[bookings["end"] > bookings["start"]].copy()
occupancy = occupancy.dropna(subset=["timestamp", "facility"]).copy()

# Fill missing occupancy readings with facility-and-hour median, then global median.
occupancy["hour"] = occupancy["timestamp"].dt.hour
occupancy["date"] = occupancy["timestamp"].dt.normalize()
occ_median = occupancy.groupby(["facility", "hour"])["count"].transform("median")
occupancy["count_was_missing"] = occupancy["count"].isna()
occupancy["count"] = occupancy["count"].fillna(occ_median)
occupancy["count"] = occupancy["count"].fillna(occupancy["count"].median()).round().astype(int)

# Flag readings over capacity; retain original value for audit, cap to capacity for utilization calculations.
capacity_map = facilities.set_index("facility")["capacity"]
occupancy["capacity"] = occupancy["facility"].map(capacity_map)
occupancy["over_capacity_flag"] = occupancy["count"] > occupancy["capacity"]
occupancy["count_original"] = occupancy["count"]
occupancy["count"] = np.minimum(occupancy["count"], occupancy["capacity"]).astype(int)

# Incomplete maintenance records: unknown issue label; missing downtime filled by median, or 0 if all missing.
maintenance["issue"] = maintenance["issue"].fillna("Unknown issue")
maintenance["downtime"] = maintenance["downtime"].fillna(maintenance["downtime"].median())
maintenance["downtime"] = maintenance["downtime"].fillna(0)
maintenance = maintenance[maintenance["facility"].isin(facilities["facility"])].copy()

print("\nAfter cleaning:")
print("Bookings:", bookings.shape, "| Occupancy:", occupancy.shape, "| Maintenance:", maintenance.shape)
print("Missing occupancy counts remaining:", int(occupancy["count"].isna().sum()))
print("Over-capacity readings flagged:", int(occupancy["over_capacity_flag"].sum()))

# STEP 6: Create combined analysis dataset at facility-hour level
occupancy["date_only"] = occupancy["timestamp"].dt.normalize()
occupancy["hour"] = occupancy["timestamp"].dt.hour
occupancy["day_of_week"] = occupancy["timestamp"].dt.day_name()
occupancy["is_weekend"] = occupancy["timestamp"].dt.weekday >= 5

calendar_small = calendar.rename(columns={"date": "date_only"})
calendar_small["date_only"] = pd.to_datetime(calendar_small["date_only"]).dt.normalize()
occupancy = occupancy.merge(
    calendar_small[["date_only", "working_day", "event", "exam_period"]],
    on="date_only", how="left"
)
occupancy = occupancy.merge(
    facilities[["facility", "type", "block", "capacity"]],
    on=["facility", "capacity"], how="left"
)
occupancy["utilization_pct"] = (occupancy["count"] / occupancy["capacity"] * 100).round(2)
occupancy["utilization_band"] = pd.cut(
    occupancy["utilization_pct"], bins=[-1, 25, 60, 85, 100],
    labels=["Low", "Moderate", "High", "Near capacity"]
)

# Booking-level engineered features
bookings["duration_hours"] = bookings["end"] - bookings["start"]
bookings["day_of_week"] = bookings["date"].dt.day_name()
bookings["is_weekend"] = bookings["date"].dt.weekday >= 5
bookings = bookings.merge(calendar[["date", "working_day", "event", "exam_period"]], on="date", how="left")

# Save cleaned files and combined dataset
facilities.to_csv(OUT / "cleaned_facilities.csv", index=False)
bookings.to_csv(OUT / "cleaned_bookings.csv", index=False)
occupancy.to_csv(OUT / "cleaned_occupancy_analysis.csv", index=False)
calendar.to_csv(OUT / "cleaned_calendar.csv", index=False)
maintenance.to_csv(OUT / "cleaned_maintenance.csv", index=False)

print("Saved cleaned CSV files.")

# STEP 7: Descriptive analysis
print("\nFacility information:")
display(facilities.describe(include="all"))
print("\nOccupancy descriptive statistics:")
display(occupancy[["count", "capacity", "utilization_pct"]].describe().round(2))

facility_usage = occupancy.groupby(["facility", "type"], as_index=False).agg(
    average_occupancy=("count", "mean"),
    peak_occupancy=("count", "max"),
    capacity=("capacity", "first"),
    average_utilization_pct=("utilization_pct", "mean"),
    peak_utilization_pct=("utilization_pct", "max"),
    readings=("count", "size")
).sort_values("average_utilization_pct", ascending=False)
print("Facility usage ranking:")
display(facility_usage.round(2))

hourly_usage = occupancy.groupby("hour", as_index=False).agg(
    average_occupancy=("count", "mean"),
    average_utilization_pct=("utilization_pct", "mean")
)
print("Hourly usage:")
display(hourly_usage.round(2))

underused = facility_usage[facility_usage["average_utilization_pct"] < 25].copy()
print("Underused facilities (average utilization < 25%):")
display(underused.round(2))

# STEP 8: Required visualizations
# Chart 1: Average utilization by facility
plt.figure(figsize=(12, 5))
plot_data = facility_usage.sort_values("average_utilization_pct", ascending=False)
sns.barplot(data=plot_data, x="facility", y="average_utilization_pct")
plt.title("Average Utilization by Facility")
plt.xlabel("Facility")
plt.ylabel("Average utilization (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Chart 2: Peak-time analysis
plt.figure(figsize=(10, 5))
sns.lineplot(data=hourly_usage, x="hour", y="average_utilization_pct", marker="o")
plt.title("Average Utilization by Hour")
plt.xlabel("Hour of day")
plt.ylabel("Average utilization (%)")
plt.xticks(range(8, 18))
plt.tight_layout()
plt.show()

# Chart 3: Exam vs non-exam utilization
exam_compare = occupancy.groupby("exam_period", as_index=False).agg(
    average_utilization_pct=("utilization_pct", "mean"),
    average_occupancy=("count", "mean")
)
exam_compare["period"] = exam_compare["exam_period"].map({True: "Exam period", False: "Non-exam period"})
plt.figure(figsize=(7, 5))
sns.barplot(data=exam_compare, x="period", y="average_utilization_pct")
plt.title("Exam Period vs Non-Exam Period")
plt.xlabel("")
plt.ylabel("Average utilization (%)")
plt.tight_layout()
plt.show()
display(exam_compare.round(2))

# Chart 4: Utilization by facility type
type_usage = occupancy.groupby("type", as_index=False)["utilization_pct"].mean().sort_values("utilization_pct", ascending=False)
plt.figure(figsize=(9, 5))
sns.barplot(data=type_usage, x="type", y="utilization_pct")
plt.title("Average Utilization by Facility Type")
plt.xlabel("Facility type")
plt.ylabel("Average utilization (%)")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()

# Chart 5: Capacity vs peak occupancy
capacity_peak = facility_usage.sort_values("capacity")
plt.figure(figsize=(9, 6))
sns.scatterplot(data=capacity_peak, x="capacity", y="peak_occupancy", hue="type", s=100)
plt.title("Facility Capacity vs Peak Occupancy")
plt.xlabel("Capacity")
plt.ylabel("Peak occupancy")
plt.tight_layout()
plt.show()

# Chart 6: Maintenance downtime by facility
downtime_summary = maintenance.groupby("facility", as_index=False).agg(
    total_downtime=("downtime", "sum"),
    maintenance_incidents=("issue", "size")
).sort_values("total_downtime", ascending=False)
plt.figure(figsize=(10, 5))
sns.barplot(data=downtime_summary, x="facility", y="total_downtime")
plt.title("Total Maintenance Downtime by Facility")
plt.xlabel("Facility")
plt.ylabel("Downtime (hours)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
display(downtime_summary)

# Chart 7: Utilization heatmap by facility and hour
heatmap_data = occupancy.pivot_table(index="facility", columns="hour", values="utilization_pct", aggfunc="mean")
plt.figure(figsize=(12, 6))
sns.heatmap(heatmap_data, annot=True, fmt=".0f", cmap="YlGnBu")
plt.title("Facility Utilization Heatmap by Hour (%)")
plt.xlabel("Hour")
plt.ylabel("Facility")
plt.tight_layout()
plt.show()

# STEP 9: Explore relationships and patterns
print("Correlation matrix:")
corr = occupancy[["count", "capacity", "utilization_pct", "hour", "exam_period", "working_day"]].corr(numeric_only=True)
display(corr.round(2))
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm", center=0, fmt=".2f")
plt.title("Correlation Between Numeric Analysis Features")
plt.tight_layout()
plt.show()

print("Utilization by exam period and facility type:")
exam_type = occupancy.groupby(["exam_period", "type"], as_index=False)["utilization_pct"].mean()
display(exam_type.pivot(index="type", columns="exam_period", values="utilization_pct").round(2))

print("Utilization by weekday:")
weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
weekday_usage = occupancy.groupby("day_of_week", as_index=False)["utilization_pct"].mean()
weekday_usage["day_of_week"] = pd.Categorical(weekday_usage["day_of_week"], categories=weekday_order, ordered=True)
weekday_usage = weekday_usage.sort_values("day_of_week")
display(weekday_usage.round(2))

# STEP 10: Maintenance frequency and capacity analysis
maintenance_summary = maintenance.groupby("facility", as_index=False).agg(
    incidents=("issue", "size"),
    total_downtime_hours=("downtime", "sum"),
    average_downtime_hours=("downtime", "mean")
).sort_values("total_downtime_hours", ascending=False)
print("Maintenance summary:")
display(maintenance_summary.round(2))

capacity_analysis = facility_usage[["facility", "type", "capacity", "average_occupancy", "peak_occupancy",
                                   "average_utilization_pct", "peak_utilization_pct"]].copy()
capacity_analysis["peak_capacity_gap"] = capacity_analysis["capacity"] - capacity_analysis["peak_occupancy"]
capacity_analysis["peak_over_capacity"] = capacity_analysis["peak_occupancy"] > capacity_analysis["capacity"]
print("Capacity comparison:")
display(capacity_analysis.round(2))

```

```python

# STEP 11: DAILY DATASET + FEATURE ENGINEERING

# Ensure dates are in the correct format
occupancy["date_only"] = pd.to_datetime(occupancy["timestamp"]).dt.normalize()
bookings["date_only"] = pd.to_datetime(bookings["date"]).dt.normalize()

# Create daily occupancy features
daily = occupancy.groupby(
    ["facility", "date_only", "type", "block", "capacity",
     "working_day", "event", "exam_period", "is_weekend"],
    as_index=False
).agg(
    avg_daily_occupancy=("count", "mean"),
    max_daily_occupancy=("count", "max"),
    avg_daily_utilization_pct=("utilization_pct", "mean"),
    peak_daily_utilization_pct=("utilization_pct", "max"),
    missing_readings_filled=("count_was_missing", "sum"),
    over_capacity_events=("over_capacity_flag", "sum")
)

daily["day_of_week"] = daily["date_only"].dt.day_name()
daily["month"] = daily["date_only"].dt.month

# Create booking features
bookings["duration_hours"] = bookings["end"] - bookings["start"]

booking_daily = bookings.groupby(
    ["facility", "date_only"], as_index=False
).agg(
    booking_count=("purpose", "size"),
    total_booked_hours=("duration_hours", "sum"),
    unique_booking_purposes=("purpose", "nunique")
)

# Merge booking features into daily dataset
daily = daily.merge(
    booking_daily,
    on=["facility", "date_only"],
    how="left"
)

booking_cols = [
    "booking_count",
    "total_booked_hours",
    "unique_booking_purposes"
]
daily[booking_cols] = daily[booking_cols].fillna(0)

print("Feature engineering completed!")
print("Daily dataset shape:", daily.shape)
display(daily.head())


# STEP 12: RANDOM FOREST CLASSIFICATION

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, f1_score, classification_report,
    confusion_matrix
)

# Define target: High utilization >= 60%, otherwise Low utilization
daily["target"] = np.where(
    daily["avg_daily_utilization_pct"] >= 60,
    "High Utilization",
    "Low Utilization"
)

# Select features available before calculating the target
features = [
    "facility", "type", "block", "capacity",
    "working_day", "event", "exam_period", "is_weekend",
    "day_of_week", "month", "booking_count",
    "total_booked_hours", "unique_booking_purposes",
    "missing_readings_filled", "over_capacity_events"
]

X = daily[features]
y = daily["target"]

# Separate numerical and categorical features
cat_cols = X.select_dtypes(
    include=["object", "category", "string"]
).columns.tolist()
num_cols = [col for col in X.columns if col not in cat_cols]

# Handle missing values and encode categories
preprocessor = ColumnTransformer([
    ("numeric", SimpleImputer(strategy="median"), num_cols),
    ("categorical", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]), cat_cols)
])

# Build Random Forest model
rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        class_weight="balanced"
    ))
])

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Train and predict
rf_model.fit(X_train, y_train)
y_pred = rf_model.predict(X_test)

# STEP 13: MODEL PERFORMANCE

print("MODEL: Random Forest Classifier")
print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print(f"Weighted F1-score: {f1_score(y_test, y_pred, average='weighted'):.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# Feature importance
encoder = rf_model.named_steps["preprocessor"]
feature_names = encoder.get_feature_names_out()
importance = rf_model.named_steps["classifier"].feature_importances_

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importance
}).sort_values("Importance", ascending=False)

print("\nTop 10 Important Features:")
display(feature_importance.head(10))

# Save outputs
daily.to_csv(OUT / "engineered_daily_facility_dataset.csv", index=False)
feature_importance.to_csv(OUT / "random_forest_feature_importance.csv", index=False)

print("Daily dataset and Random Forest results saved.")
```
